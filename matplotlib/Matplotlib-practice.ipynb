```python
#for using catagorical values we cannot use histogram or line chart
#pie chart
count=df["dept"].value_counts()
count
```

```python
plt.pie(count)
plt.show()
```

```python
#in above cannot identify what color representing what so we use labels=count.index
plt.pie(count,labels=count.index)
plt.show()
```

```python
#again in above we cannot see values or percent so we use autopct="%1.1f"
plt.pie(count,labels=count.index,autopct="%1.1f")# we used 1.1 because it give .1 decimal if we give 2 it will give two decimal
plt.show()
```

```python
plt.pie(count,labels=count.index,autopct="%1.1f",explode=[0, 0.1, 0, 0, 0, 0, 0])# we can also use explode to saparete any one point explode=[0,0.1,0,0,0,0,0]
plt.show()
```

```python
#countplot
plt.bar(count.index,count,)
plt.show()
```

```python
plt.bar(count.index,count,color=["green","red","black","yellow",])
plt.show()
```

```python
#Line plot
axs[0].plot(sort_age["age"],df["salary"],color="red",marker="*",linewidth="2",markersize=2)
axs[0].grid()
axs[0].set_title("Line plot")
axs[0].set_xlabel("age")
axs[0].set_ylabel("salary")

#histogram
axs[1].hist(df["salary"],bins=5,color="skyblue")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

#box plot
axs[2].boxplot(df["salary"])
axs[2].set_title("box plot")
axs[2].set_xlabel("salary")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Create 3 subplots
fig, axs = plt.subplots(1, 3, figsize=(15, 5))

# Line plot
axs[0].plot(sort_age["age"], sort_age["salary"],
            color="red", marker="*", linewidth=2, markersize=8)
axs[0].grid()
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

# Histogram
axs[1].hist(df["salary"], bins=5, color="skyblue")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

# Box plot
axs[2].boxplot(df["salary"])
axs[2].set_title("Box Plot")
axs[2].set_xlabel("Salary")

# Display all graphs
plt.tight_layout()
plt.savefig("multipleplots.png")
plt.show()
```
